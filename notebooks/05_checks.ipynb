```python
import pandas as pd
t = pd.read_csv("../data/covid_trend.csv", parse_dates=["date"])
c = t[t["country"] == "Chile"].set_index("date").sort_index()
print(c["new_deaths_smoothed_per_million"].nlargest(8))
print(c.loc["2022-02-15":"2022-04-15", ["new_deaths_smoothed_per_million", "total_deaths_per_million"]].iloc[::3])
print("Chile total deaths per million at end of data:", c["total_deaths_per_million"].dropna().iloc[-1])
```

```python
g = t.groupby("country")["new_deaths_smoothed_per_million"]
chk = pd.DataFrame({"peak": g.max(), "p99": g.quantile(0.99)})
chk["ratio"] = chk["peak"] / chk["p99"]
print(chk.sort_values("ratio", ascending=False).head(10).round(1))
```
